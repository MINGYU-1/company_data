##### JOIN TEST

```python
# 분석용 기본 세팅
import pandas as pd
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:fls0413!@localhost:3306/dashboard")
```

```python
query = """
SELECT r.사업장명, b.정류장명, n.거리
FROM near_bus_stops n
JOIN restaurants r ON n.사업장명 = r.사업장명
JOIN bus_stops b ON n.정류장ID = b.정류장
LIMIT 20;
"""

df_join = pd.read_sql(query, engine)
df_join
```

##### 음식점별 가까운 정류장 

```python
df_min_dist = pd.read_sql("""
SELECT 사업장명, MIN(거리) AS 최소거리
FROM near_bus_stops
GROUP BY 사업장명;
""", engine)

df_min_dist.head()
```

#### 지역별 평균 접근성 

```python
df_region = pd.read_sql("""
SELECT r.sigungu, AVG(n.거리) AS 평균거리
FROM near_bus_stops n
JOIN restaurants r ON n.사업장명 = r.사업장명
GROUP BY r.sigungu;
""", engine)

df_region
```

#### 지도 시각화용 데이터 

```python
df_map = pd.read_sql("""
SELECT 
    r.사업장명, r.lat AS 업소_lat, r.lng AS 업소_lng,
    b.정류장명, b.위도 AS 정류_lat, b.경도 AS 정류_lng,
    n.거리
FROM near_bus_stops n
JOIN restaurants r ON n.사업장명 = r.사업장명
JOIN bus_stops b ON n.정류장ID = b.정류장;
""", engine)

df_map.head()
```
