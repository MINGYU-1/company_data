```python
import os
import pandas as pd

folder = "./data_csv"  # 상대경로

# 시도해 볼 인코딩 후보들
encoding_candidates = ["cp949", "euc-kr", "utf-8-sig", "utf-8"]

for filename in os.listdir(folder):
    if filename.lower().endswith(".csv"):
        path = os.path.join(folder, filename)

        df = None
        last_error = None

        for enc in encoding_candidates:
            try:
                df = pd.read_csv(path, encoding=enc)
                # 성공하면 enc 정보 출력해도 좋음
                print(f"{filename} (encoding={enc}) → row 수: {len(df)}")
                break
            except UnicodeDecodeError as e:
                last_error = e
                continue

        if df is None:
            print(f"{filename} 은(는) 인코딩 문제로 읽지 못했습니다.")
            if last_error is not None:
                print("  마지막 오류:", last_error)
```
