```python
import pandas as pd
data = pd.read_csv("./data_analysis/summary_sorted.csv")
data.drop(columns = ['encoding','Unnamed: 0'],inplace = True)
data.info()
```

```python
data.sort_values(by ='rows', ascending= False, inplace= True)
```

```python
data = data.head(10)
data.rename(columns = {'filename':'업종','rows':'사업현황'},inplace = True)
```

```python
data.columns
```

```python
data.set_index('업종')
```

```python
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'   # ★ 한국어 폰트 지정
plt.rcParams['axes.unicode_minus'] = False      # - 기호 깨짐 방지

ax = data['사업현황'].plot.bar(figsize=(12,6))

new_labels = [idx.split('_', 1)[1] for idx in data.index]

ax.set_xticklabels(new_labels, rotation=45, ha='right')

ax.set_xlabel('업종')
ax.set_ylabel('사업현황')

plt.tight_layout()
plt.show()
```
