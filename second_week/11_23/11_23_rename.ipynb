```python
import os
import re

# CSV 파일이 있는 폴더 경로 (네 환경에 맞게 수정)
folder ="./data_csv"
# 파일 목록 불러오기
for filename in os.listdir(folder):
    # CSV만 처리
    if filename.endswith(".csv"):
        
        # 정규식: 숫자 + 언더바 제거 (ex: 6530000_ → "")
        new_name = re.sub(r"^\d+_", "", filename)

        # 원본 경로 / 새 경로
        old_path = os.path.join(folder, filename)
        new_path = os.path.join(folder, new_name)

        # 실제 파일명 변경
        os.rename(old_path, new_path)

        print(f"변경됨 → {filename}  →  {new_name}")

print("🎉 모든 CSV 파일 이름 변경 완료!")
```

```python
import os
import re

folder = "./data_csv"  # 너의 실제 경로로 바꿔

for filename in os.listdir(folder):
    if filename.endswith(".csv"):

        # 1) 앞쪽 숫자 구간 제거: _01_01_01_ → _
        temp = re.sub(r"_\d+(?:_\d+)*_", "_", filename)

        # 2) 남아 있는 '_P_' 구간 제거
        new_name = temp.replace("_P_", "_")

        old_path = os.path.join(folder, filename)
        new_path = os.path.join(folder, new_name)

        os.rename(old_path, new_path)
        print(f"{filename} → {new_name}")
```

```python
import os, unicodedata

path = './data_csv'

for f in os.listdir(path):
    nfkc = unicodedata.normalize('NFC', f)
    if f != nfkc:
        os.rename(os.path.join(path, f), os.path.join(path, nfkc))
        print("변경됨:", f, "→", nfkc)
```
