```python
import os
import pandas as pd

# CSV 폴더 경로
folder = "/Users/mac/Desktop/project/company_data/second_week/data_csv"

# 폴더 내 CSV 파일 목록 가져오기
csv_files = [f for f in os.listdir(folder) if f.endswith(".csv")]

print(f"총 CSV 파일 수: {len(csv_files)}개")

# 모든 CSV의 컬럼명 저장 리스트
all_columns = []

for file in csv_files:
    path = os.path.join(folder, file)
    try:
        df = pd.read_csv(path, encoding="cp949")
        all_columns.append(set(df.columns))
        print(f"불러옴: {file} → {len(df.columns)} columns")
    except:
        print(f"⚠️ 불러오기 실패: {file}")

# =========================
#  📌 모든 CSV의 공통 컬럼(교집합)
# =========================
common_columns = set.intersection(*all_columns)

print("\n===== ✅ 모든 CSV의 '공통 컬럼' 목록 =====")
for col in sorted(common_columns):
    print(col)

print(f"\n총 공통 컬럼 수: {len(common_columns)}개")
```

```python
import os
import pandas as pd

folder = "../data_csv"

common_cols = None
file_cols = {}
col_has_value = {}   # 각 컬럼이 어떤 파일에서라도 값이 있었는지 기록

for f in os.listdir(folder):
    if f.endswith(".csv"):
        try:
            df = pd.read_csv(
                os.path.join(folder, f),
                encoding='cp949',
                on_bad_lines='skip'
            )
            file_cols[f] = set(df.columns)

            # 공통 컬럼 계산
            if common_cols is None:
                common_cols = set(df.columns)
            else:
                common_cols = common_cols.intersection(df.columns)

            # 값이 있는 컬럼 기록
            for col in df.columns:
                # NaN이 아닌 값이 하나라도 있는지
                has_val = df[col].notna().any()
                col_has_value[col] = col_has_value.get(col, False) or has_val

        except Exception as e:
            print(f"❌ Error reading {f}: {e}")

# ------------------------
# 모든 파일에서 공통이지만, 값이 단 한 번도 없었던 컬럼 제외
# ------------------------
final_common_cols = {col for col in common_cols if col_has_value.get(col, False)}

print("\n📌 null이 아닌 값이 하나라도 있는 공통 컬럼 목록:")
print(final_common_cols)

print(f"\n총 공통 컬럼 수: {len(final_common_cols)}")
```
