## 컴럼명에 대한 dataFrame제작


```python
import pandas as pd
from pathlib import Path

# --- 1. 경로 설정 (이전 단계에서 설정된 값 사용) ---
current_path = Path.cwd()
parent_path = current_path.parent
folder_path = parent_path / 'second_week' / 'data_csv'
output_filename = 'csv_column_summary.csv' # 최종 저장될 CSV 파일 이름
output_path = current_path / output_filename

print(f"--- 대상 폴더: {folder_path} ---")

# 모든 파일의 컬럼 정보를 저장할 리스트 초기화
column_summaries = []

if not folder_path.is_dir():
    print(f"오류: 타겟 폴더를 찾을 수 없거나 디렉토리가 아닙니다: {folder_path}")
else:
    all_files = list(folder_path.glob('*.csv'))

    if not all_files:
        print("대상 폴더에서 CSV 파일을 찾을 수 없습니다.")
    
    # --- 2. 파일 순회 및 결과 수집 (기존 try-except 로직 활용) ---
    for file in all_files:
        print(f"\n[파일 처리 시작] {file.name}")
        
        # 기본 컬럼명 결과를 None으로 초기화
        extracted_columns = None
        error_message = None

        try:
            df = pd.read_csv(file, encoding='cp949')
            column_names = df.columns.to_list() 
            
            extracted_columns = str(column_names) # 리스트를 문자열로 저장
            print(f"  ✅ 컬럼명 추출 성공: {file.name}")
            
        except FileNotFoundError:
            error_message = "파일을 찾을 수 없습니다."
            print(f"  ❌ 오류: {error_message} (경로: {file})")
            
        except UnicodeDecodeError:
            error_message = "인코딩 문제 발생. cp949로 디코딩 불가."
            print(f"  ❌ 오류: {error_message}")
            
        except pd.errors.EmptyDataError:
            error_message = "파일이 비어있습니다 (데이터 없음)."
            print(f"  ❌ 오류: {error_message}")

        except Exception as e:
            error_message = f"예기치 않은 오류 발생: {type(e).__name__}: {e}"
            print(f"  ❌ 오류: {error_message}")
        
        # 파일 처리 결과를 리스트에 저장
        column_summaries.append({
            'FileName': file.name,
            'Columns': extracted_columns if extracted_columns is not None else 'N/A',
            'Error_Status': error_message if error_message is not None else 'Success'
        })
        
    # --- 3. 수집된 결과를 DataFrame으로 변환 및 CSV로 저장 ---
    
    # 리스트를 Pandas DataFrame으로 변환
    summary_df = pd.DataFrame(column_summaries)
    
    # CSV 파일로 저장
    try:
        summary_df.to_csv(output_path, index=False, encoding='utf-8')
        print(f"\n=======================================================")
        print(f"🎉 모든 파일의 컬럼 정보를 성공적으로 저장했습니다.")
        print(f"저장 경로: {output_path}")
        print(f"=======================================================")
    except Exception as e:
        print(f"\n최종 요약 CSV 저장 중 오류 발생: {e}")

```
