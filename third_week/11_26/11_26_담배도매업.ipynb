#### 시군별 폐업률
##### 폐업률 말이 안 되는데.. 어디서 잘못한 건지 모르겠어요..

```python
total = df.groupby("시군")["시군"].count()
closed = df[df["폐업일자"].notna()].groupby("시군")["폐업일자"].count()
rate = (closed/total * 100).fillna(0)
df1 = rate.reset_index(name="폐업률")
df1
```
