```python
nb_name = ['미용업', '즉석판매제조가공업', '농어촌민박업', '통신판매업', '담배도매업', '건강기능식품유통전문판매업', '가축사육업', '일반음식점', '휴게음식점']
csv_folder = "/Users/mac/Desktop/project/company_data/second_week/data_csv"
csv_files = glob.glob(os.path.join(csv_folder,"*.csv"))

target_csv = []

for nb in nb_name:
    for csv in csv_files:
        if nb in csv:
            target_csv.append(csv)
target_csv = [f for f in target_csv if "동물미용업" not in f]
target_csv
```

#### 업종명마다 어떤 업태구분명이 있는지 확인함

```python
rows = []
for csv in target_csv:
    df = pd.read_csv(csv, encoding= "CP949", on_bad_lines="skip")
    df.columns=df.columns.str.strip()

    if "개방서비스명" in df.columns and "업태구분명" in df.columns:
        업종명 = csv.split("_")[-1].replace(".csv","")

        for svc, up in zip(df["개방서비스명"], df["업태구분명"]):
            rows.append([업종명, svc, up])
cols = ["업종명", "개방서비스명", "업태구분명"]
result = pd.DataFrame(rows, columns=["업종명", "개방서비스명", "업태구분명"])
result
```
