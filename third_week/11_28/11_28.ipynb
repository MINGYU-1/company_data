```python
from pathlib import Path
import os
import pandas as pd

current_path = Path.cwd()
parent_path = current_path.parent
file_path = parent_path/'second_week/data_csv'
file = file_path/'강원특별자치도_일반음식점.csv'
df = pd.read_csv(file,encoding = 'cp949')
df1 = df.copy()
```

```python
df1['폐업유무'] = df1['폐업일자'].isna()
```

```python
df1.info()
```

```python
# 1. '호프/통닭'을 '통닭(치킨)'으로 변경하여 두 값을 '통닭(치킨)'으로 통일
df1.loc[df1['위생업태명'] == '통닭(치킨)', '위생업태명'] = '호프/통닭'

# 결과 확인: '호프/통닭'은 사라지고 '통닭(치킨)'의 개수가 증가함
print(df1['위생업태명'].value_counts())
```

```python
columns = ['소재지면적','폐업유무']
df2 = df1[columns]
df2
columns_to_check = ['소재지면적', '폐업유무']
df_cleaned = df2.dropna(subset=columns_to_check)
df_cleaned.info()
```

```python
# df_x는 '소재지면적' 컬럼의 데이터를 NumPy 배열로 가져와 (n, 1) 형태의 2차원 배열로 만듭니다.

df_x = df_cleaned['소재지면적'].to_numpy().reshape(-1, 1)
df_y = df_cleaned['폐업유무'].to_numpy().reshape(-1,1)
df_x.shape
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(df_x,df_y,random_state=5, test_size=0.5)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
y_predict = model.predict(x_train)
print(model.score(x_train,y_train))
print(model.score(x_test,y_test))
```
