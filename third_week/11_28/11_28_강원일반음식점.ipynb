```python
import pandas as pd
df = pd.read_csv("/Users/mac/Desktop/project/company_data/third_week/data_csv/강원일반음식점.csv", encoding="CP949")
```

```python
df["폐업여부"] = df["폐업일자"].notnull().astype(int)
df["소재지면적"] = df["소재지면적"].fillna(df["소재지면적"].median())
```

```python
df = pd.get_dummies(df, columns=["위생업태명"], drop_first=True)
```

```python
use_cols = ["소재지면적"] + [c for c in df.columns if "위생업태명_" in c]
x = df[use_cols]
y = df["폐업여부"]

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

model = LogisticRegression(max_iter=2000)
model.fit(x_train, y_train)
```

```python
coeff = pd.DataFrame({"feature": x_train.columns, "coef": model.coef_[0]}).sort_values("coef", ascending=False)
coeff
```
